```python
import numpy as np
from pyhdf.SD import SD, SDC
import scipy.ndimage
import matplotlib.pyplot as plt

# --- User-Defined Parameters ---
# Path to your manually downloaded HDF file
file_path = r'/mnt/c/Users/vvagh/OneDrive - Indian Institute of Science Education and Research Bhopal/Documents/ML-hands-on/MYD06_L2.A2003141.2355.061.2018007051650.hdf'

# Define the geographic bounding box (lat_min, lat_max, lon_min, lon_max).
# For example, continental USA:
LAT_MIN, LAT_MAX = 24.396308, 49.384358
LON_MIN, LON_MAX = -125.0, -66.93457

# --- Open the HDF file ---
try:
    hdf = SD(file_path, SDC.READ)
except Exception as e:
    print(f"Error opening file {file_path}: {e}")
    exit(1)

# --- Extract the 1-km Cloud Mask ---
try:
    # "Cloud_Mask_1km" is typically stored with dimensions: (2030, 1354, 2).
    cloud_mask = hdf.select('Cloud_Mask_1km').get()
    # Use only the first band for demonstration.
    cloud_mask = cloud_mask[:, :, 0]  # shape should be (2030, 1354)
    print(f"Cloud_Mask_1km shape: {cloud_mask.shape}")
except Exception as e:
    print(f"Error reading 'Cloud_Mask_1km': {e}")
    exit(1)

# --- Extract 5-km Geolocation Arrays ---
try:
    lat_5km = hdf.select('Latitude').get()   # shape (406, 270)
    lon_5km = hdf.select('Longitude').get()    # shape (406, 270)
    print(f"Latitude shape (5km): {lat_5km.shape}")
    print(f"Longitude shape (5km): {lon_5km.shape}")
except Exception as e:
    print(f"Error reading geolocation arrays: {e}")
    exit(1)

# --- Upsample the 5-km Geolocation Arrays to the 1-km Grid ---
row_zoom = cloud_mask.shape[0] / lat_5km.shape[0]   # ~2030/406 = 5.0
col_zoom = cloud_mask.shape[1] / lon_5km.shape[1]     # ~1354/270 ≈ 5.01

lat_1km = scipy.ndimage.zoom(lat_5km, zoom=(row_zoom, col_zoom), order=0)
lon_1km = scipy.ndimage.zoom(lon_5km, zoom=(row_zoom, col_zoom), order=0)
print(f"Upsampled Latitude shape (1km): {lat_1km.shape}")
print(f"Upsampled Longitude shape (1km): {lon_1km.shape}")

# Print min and max to inspect geographic coverage
print("Latitude range (1km):", np.min(lat_1km), np.max(lat_1km))
print("Longitude range (1km):", np.min(lon_1km), np.max(lon_1km))

# --- Apply the Bounding Box to Crop the 1-km Cloud Mask Data ---
mask = (
    (lat_1km >= LAT_MIN) & (lat_1km <= LAT_MAX) &
    (lon_1km >= LON_MIN) & (lon_1km <= LON_MAX)
)
cloud_mask_subset = cloud_mask[mask]
print(f"Cropped cloud mask shape: {cloud_mask_subset.shape}")

if cloud_mask_subset.size == 0:
    print("No cloud mask data in the defined bounding box.")

# --- Visualization ---
# Plot the upsampled geolocation grid and overlay the bounding box.
plt.figure(figsize=(10, 6))
plt.scatter(lon_1km.flatten(), lat_1km.flatten(), s=1, c='blue', label="Geolocation points")
# Draw bounding box edges
plt.plot([LON_MIN, LON_MAX, LON_MAX, LON_MIN, LON_MIN],
         [LAT_MIN, LAT_MIN, LAT_MAX, LAT_MAX, LAT_MIN],
         'r-', lw=2, label="Bounding Box")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Upsampled Geolocation Data with Bounding Box")
plt.legend()
plt.grid(True)
plt.show()

# Optionally, visualize the cloud mask data (using imshow)
plt.figure(figsize=(10, 6))
plt.imshow(cloud_mask, cmap='gray')
plt.title("Cloud Mask 1km (Raw)")
plt.colorbar(label="Cloud Mask Value")
plt.show()
```
